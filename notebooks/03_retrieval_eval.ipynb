# Phase 2 Evaluation: Retrieval System

This notebook documents the performance of the FAISS retrieval system using `all-MiniLM-L6-v2` embeddings over the OpenI radiology reports.

## 1. Quantitative Metrics (Validation Set)

We evaluate the **Top-k Hit Rate**: for a given report query, we check if any of the top-k retrieved chunks belong to the same study.

| Metric | Value |
|---|---|
| **Hit@1** | 0.1348 |
| **Hit@3** | 0.2222 |
| **Hit@5** | 0.2459 |
| **Random Baseline** | 0.000045 |

**Conclusion:** The retriever significantly outperforms the random baseline, providing a solid foundation for grounded generation.
